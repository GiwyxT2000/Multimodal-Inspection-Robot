# 智能体编排器

完整流程：定位牌停车 -> 云台搜索工况牌 -> 小车转向工况牌 -> 声光风险反馈。

```python
from inspection_agent_orchestrator import run_demo, summarize_report

OUTPUT_DIR = "inspection_agent_output"
COMMAND = "开始A01工位巡检"
# None 表示使用视觉模块自己的默认模型。
MODEL = None

# 第一次建议 False：锚点阶段和车身转向都是 dry run。
# 搜索和声光反馈仍会执行。
ENABLE_MOTION = False

# 如果你想先跳过直线靠近定位牌，只测试搜索->转向->反馈，可以设为 False。
RUN_ANCHOR = True

RUN_SEARCH = True
RUN_MOTION = True
RUN_FEEDBACK = True

# 默认不覆盖各模块参数；各模块会使用自己的默认值。
# 若临时需要覆盖，可在 run_demo(...) 中显式传入对应参数。
SEARCH_PROFILE = None
PAN_DELTA = None
```

```python
report = run_demo(
    output_dir=OUTPUT_DIR,
    command=COMMAND,
    model=MODEL,
    enable_motion=ENABLE_MOTION,
    run_anchor=RUN_ANCHOR,
    run_search=RUN_SEARCH,
    run_motion=RUN_MOTION,
    run_feedback=RUN_FEEDBACK,
    require_anchor_success=True,
    require_target_found=True,
    search_profile=SEARCH_PROFILE,
    pan_delta=PAN_DELTA,
)
print(summarize_report(report))
```
