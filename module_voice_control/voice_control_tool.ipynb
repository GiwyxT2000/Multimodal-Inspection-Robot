# Voice Control Tool

独立测试 Yahboom AI 语音模块：监听固定命令词触发巡检，并按风险等级发送固定播报词。

```python
from voice_control_tool import run_demo

# 常用配置
PORT = None          # None 表示自动搜索 /dev/ttyUSB* 或 /dev/ttyACM*
LISTEN_SECONDS = 30
RISK_LEVEL = "abnormal"  # normal / warning / abnormal
```

## 1. Dry Run

不需要连接语音模块，只检查协议帧和报告输出。

```python
report = run_demo(
    output_dir="voice_control_output_dryrun",
    port=PORT,
    risk_level=RISK_LEVEL,
    dry_run=True,
)
report
```

## 2. 播报测试

连接语音模块后运行。会根据风险等级播报固定短句。

```python
report = run_demo(
    output_dir="voice_control_output_broadcast",
    port=PORT,
    risk_level=RISK_LEVEL,
    broadcast_only=True,
)
report
```

## 3. 语音触发测试

先说“你好小亚”唤醒，再说“小车前进”。如果成功，`triggered=True`，`action=start_inspection`。

```python
report = run_demo(
    output_dir="voice_control_output_listen",
    port=PORT,
    listen_seconds=LISTEN_SECONDS,
    init_broadcast=True,
    ready_broadcast=True,
)
report
```

## 4. 原始串口监听

如果听得到播报，但说命令没有触发，运行这一格。它不解析协议，只记录语音模块有没有发任何串口字节回来。

```python
report = run_demo(
    output_dir="voice_control_output_raw",
    port=PORT,
    listen_seconds=LISTEN_SECONDS,
    init_broadcast=False,
    ready_broadcast=False,
    raw_monitor=True,
)
report
```
