测试 Jetbot 是否可以沿直线接近 `A01 工位定位锚点牌`，并在二维码达到设定大小后停车。

运行前请确认：

- 前方路径清空。
- 工位定位牌竖直放在正前方。
- 手放在电源开关附近，发现异常立即断电。
- 关闭其他占用相机的 Notebook kernel。

```python
from qr_anchor_approach_demo import run_demo, summarize_report

report = run_demo(
    output_dir='qr_anchor_approach_output',
    prepare=True,
    init_gimbal=True,
    require_gimbal=True,
    home_pan=2048,
    home_tilt=2868,
    gimbal_hold_seconds=1.0,
    enable_motion=True,
    camera_width=640,
    camera_height=480,
    stop_width_ratio=0.12,
    step_speed=0.55,
    step_duration=0.90,
    fine_step_speed=0.42,
    fine_step_duration=0.40,
    fine_width_ratio=0.55,
    motor_kick_speed=0.75,
    motor_kick_duration=0.12,
    max_steps=16,
    max_missing_steps=16,
    frames_per_step=5,
    post_stop_settle=0.90,
    pre_capture_settle=0.20,
    capture_sample_interval=0.12,
    capture_warmup_frames=2,
    save_frames=True,
)

print(summarize_report(report))
```
