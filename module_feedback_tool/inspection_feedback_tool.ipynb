# 声光反馈工具

本 Notebook 只测试蜂鸣器和 RGB 灯带。不会移动小车，也不会调用相机、云台或大模型。

```python
from inspection_feedback_tool import run_demo, summarize_report

OUTPUT_DIR = "inspection_feedback_output"

# normal: 绿灯 + 1声；warning: 黄灯 + 2声；abnormal: 红灯 + 3声
RISK_LEVEL = "abnormal"

# 也可以从目标搜索报告读取 risk_level：
# SEARCH_REPORT_PATH = "../module_target_search/target_search_output/target_search_report.json"
SEARCH_REPORT_PATH = None

ENABLE_BUZZER = True
ENABLE_RGB = True

# 以下为可选覆盖参数。None 表示使用 inspection_feedback_tool.py 中的模块默认值。
BEEP_ON = None
BEEP_OFF = None
RGB_LOOPS = None
RGB_BLINK_COUNT = None
```

```python
kwargs = {}
if BEEP_ON is not None:
    kwargs["beep_on"] = BEEP_ON
if BEEP_OFF is not None:
    kwargs["beep_off"] = BEEP_OFF
if RGB_LOOPS is not None:
    kwargs["rgb_loops"] = RGB_LOOPS
if RGB_BLINK_COUNT is not None:
    kwargs["rgb_blink_count"] = RGB_BLINK_COUNT

report = run_demo(
    output_dir=OUTPUT_DIR,
    risk_level=RISK_LEVEL,
    search_report_path=SEARCH_REPORT_PATH,
    enable_buzzer=ENABLE_BUZZER,
    enable_rgb=ENABLE_RGB,
    **kwargs
)
print(summarize_report(report))
```
