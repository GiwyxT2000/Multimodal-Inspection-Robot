# 小车运动工具

本 Notebook 只测试底盘运动：前进、后退、原地左转、原地右转，以及根据 `best_pose` 做车身微转。不会调用相机、云台、大模型、蜂鸣器或RGB灯带。

```python
from robot_motion_tool import run_demo, summarize_report

OUTPUT_DIR = "robot_motion_output"

# 第一次建议保持 False，确认报告正常后再改 True。
ENABLE_MOTION = True

# 根据目标搜索模块输出填写，例如 center / left / right / left_wide / right_wide。
BEST_POSE = "right_wide"

TURN_SPEED = 0.45
TURN_DURATION = 0.8
KICK_SPEED = 0.75
KICK_DURATION = 0.08
```

```python
# 根据 best_pose 让车身向目标方向微转。
report = run_demo(
    output_dir=OUTPUT_DIR,
    action="align",
    best_pose=BEST_POSE,
    enable_motion=ENABLE_MOTION,
    turn_speed=TURN_SPEED,
    turn_duration=TURN_DURATION,
    kick_speed=KICK_SPEED,
    kick_duration=KICK_DURATION,
)
print(summarize_report(report))
```
