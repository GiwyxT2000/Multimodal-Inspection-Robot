# StepFun 多模态视觉识别工具

本 Notebook 只测试视觉理解能力：读取一张已经拍好的图片，调用 `step-3.7-flash`，输出结构化 JSON 报告。

```python
# 如果你不想设置环境变量，可以临时填在这里。
# 推荐测试通过后再改用 STEPFUN_API_KEY 环境变量或 stepfun_api_key.txt。
API_KEY = ""

# 默认读取相机采集模块刚保存的稳定图片。
IMAGE_PATH = "../module_camera_capture/camera_capture_output/test_capture_best.jpg"

TASK_PROMPT = (
    "请分析这张Jetbot工业巡检图片。判断是否存在工位定位牌或工业工况巡检牌，"
    "读取可见文字，识别设备状态或安全风险，并给出简短处置建议。"
)

MODEL = "step-3.7-flash"
OUTPUT_DIR = "stepfun_vision_output"
```

```python
from stepfun_vision_tool import run_demo, summarize_report

report = run_demo(
    image_path=IMAGE_PATH,
    output_dir=OUTPUT_DIR,
    task_prompt=TASK_PROMPT,
    api_key=API_KEY or None,
    model=MODEL,
    reasoning_effort="low",
    temperature=0.0,
    max_tokens=1000,
)
print(summarize_report(report))
```

```python
# 查看结构化结果
report.get("vision", {}).get("parsed", {})
```
