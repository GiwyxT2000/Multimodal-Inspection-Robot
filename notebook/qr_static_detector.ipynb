# Static OpenCV QR Detector

静态二维码识别稳定性测试：不运动、不调用大模型，只初始化云台和相机，连续采样并统计 OpenCV QR 识别率。

```python
from qr_static_detector import run_demo, summarize_report

# 手动把小车放到对应距离后修改标签，例如 "120cm"、"100cm"、"80cm"、"50cm"。
DISTANCE_LABEL = "45cm"

report = run_demo(
    output_dir="qr_static_detector_output_" + DISTANCE_LABEL,
    distance_label=DISTANCE_LABEL,
    prepare=False,
    init_gimbal=True,
    require_gimbal=False,
    home_pan=2048,
    home_tilt=2868,
    gimbal_hold_seconds=1.0,
    camera_width=640,
    camera_height=480,
    stop_width_ratio=0.12,
    qr_real_width_mm=90.0,
    focal_px=None,
    target_role="stop_anchor",
    target_station="A01",
    sample_count=12,
    interval_seconds=0.25,
)

print(summarize_report(report))
```

```python
import json

print("Best sample:")
print(json.dumps(report.get("best_sample", {}), ensure_ascii=False, indent=2))

print("\nSample summary:")
for s in report.get("samples", []):
    d = s.get("best_detection", {})
    print(s.get("index"), "ok=", s.get("ok"), "target=", s.get("target_ok"),
          "method=", d.get("method"), "width=", d.get("qr_width_px"),
          "sharpness=", s.get("sharpness"))
```

```python
from IPython.display import Image, display

best = report.get("best_sample") or {}
if best.get("annotated_image_path"):
    print(best["annotated_image_path"])
    display(Image(filename=best["annotated_image_path"]))
```
