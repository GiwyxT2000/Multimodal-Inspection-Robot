# Camera Capture Tool

相机稳定拍照工具：初始化相机，可选云台归正，连续采样并保存最清晰的一帧。

```python
from camera_capture_tool import run_demo, summarize_report

report = run_demo(
    output_dir="camera_capture_output",
    filename_prefix="test_capture",
    init_gimbal=True,
    require_gimbal=False,
    home_pan=2048,
    home_tilt=2868,
    gimbal_hold_seconds=1.0,
    camera_width=640,
    camera_height=480,
    pre_capture_settle=0.50,
    sample_count=5,
    sample_interval=0.12,
    warmup_frames=2,
    save_all_samples=False,
)

print(summarize_report(report))
```
